```python
# pip install matplotlib

import matplotlib.pyplot as plt
```

```python
plt.plot(data['student_id'], data['total_score'])
plt.xlabel('STUDENT ID')
plt.ylabel('TOTAL SCORE')
plt.title('Student ID vs Total Score')
plt.show()
```

```python
plt.plot(data["student_id"], data["math_score"], label="Math", color='r')
plt.plot(data["student_id"], data["science_score"], label="Science", color='b')
plt.plot(data["student_id"], data["english_score"], label="English", color='g')

plt.xlabel("Student ID")
plt.ylabel("Score")
plt.title("Student Scores Comparison")

plt.legend()

plt.show()
```

```python
plt.bar(data["student_id"], data["math_score"], label="Math", color='r')
plt.show()
```

```python
plt.scatter(data["study_hours"], data["total_score"])

plt.xlabel("Study Hours")
plt.ylabel("Total Score")
plt.title("Study Hours vs Total Score")

plt.show()
```

```python
city_count = data["city"].value_counts()

plt.pie(
    city_count,
    labels=city_count.index,
    autopct="%1.1f%%"
)

plt.title("Student Distribution by City")

plt.show()
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.hist(data["math_score"])
plt.title("Math Score")

plt.subplot(2, 2, 2)
plt.hist(data["science_score"])
plt.title("Science Score")

plt.subplot(2, 2, 3)
plt.hist(data["english_score"])
plt.title("English Score")

plt.subplot(2, 2, 4)
plt.hist(data["total_score"])
plt.title("Total Score")

plt.tight_layout()

plt.show()
```

```python
# Top 10 Students

top_students = data.sort_values(
    "total_score",
    ascending=False
).head(10)

print(top_students[["name", "total_score"]])
```

```python
plt.bar(
    top_students["name"],
    top_students["total_score"]
)

plt.xlabel("Student")
plt.ylabel("Total Score")
plt.title("Top 10 Students")

plt.xticks(rotation=45)

plt.show()
```
