```python
import matplotlib.pyplot as plt
```

```python
plt.bar(data['employee_id'],data['attendance'])
plt.show()
```

```python
plt.plot(data['employee_id'],data['attendance'])
plt.show()
```

```python
plt.barh(data['employee_id'],data['attendance'])
plt.show()
```

```python
department_count = data["department"].value_counts()

plt.figure(figsize=(4, 4))

plt.pie(
    department_count,
    labels=department_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Employees by Department")
plt.show()
```

```python
gender_count = data["gender"].value_counts()

plt.figure(figsize=(4, 4))

plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Employees by Gender")
plt.show()
```
